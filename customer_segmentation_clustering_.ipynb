```python
from sklearn.cluster import KMeans
```

```python
clustering1 = KMeans(n_clusters=5)
```

```python
clustering1.fit(df_customers[['Annual Income (k$)', 'Spending Score (1-100)']])
```

```python
clustering1
```

```python

clustering1.labels_
```

```python
df_customers['income&spending cluster']=clustering1.labels_
df_customers.head(4)
```

```python
df_customers['income&spending cluster'].value_counts()
```

```python
clustering1.inertia_
```

```python
inertia_score=[]
for i in range(1,11):
  Kmeans=KMeans(n_clusters=i)
  Kmeans.fit(df_customers[['Annual Income (k$)','income&spending cluster']])
  inertia_score.append(Kmeans.inertia_)
```

```python
inertia_score
```

```python
fig, ax = plt.subplots()

ax.plot(range(1, 11), inertia_score, 'bo-')
ax.set_xlabel('number of clusters')
ax.set_ylabel('total within sum of squares')

plt.show()
```
